# Visualisation des données

## Données des utilisateurs

```python
import pandas as pd

users = pd.read_csv("users.csv")
users.head(5)
```

## Données des commandes

```python
orders = pd.read_csv("orders.csv")

orders.head(5)
```

## Vérfication de la relation entre les tables

```python
id_users_in_table_user= users['user_id'].unique().tolist()
id_users_in_table_order= orders['user_id'].unique().tolist()

# comparaison des deux listes pour trouver les IDs d'utilisateurs présents dans la table "orders" mais pas dans la table "users"
missing_users = set(id_users_in_table_order) - set(id_users_in_table_user)  

print("IDs d'utilisateurs présents dans la table 'orders' mais pas dans la table 'users':", missing_users)
```

### Vérfification s'il sagit d'un décalage ou d'un utilisateur manquant 

```python
print("id max dans la table users : ", users['user_id'].max())
print("id max dans la table orders : ", orders['user_id'].max())
```

### il y a des commandes avec un user_id=0 alors qu'il n'esiste pas dans la table users. Je supprime donc les lignes de commandes ayant un id_user=0 pour éviter des erreurs de relation lors du remplissage des tables


```python
# Supprimer les lignes où la valeur est 0 dans la colonne "id_user" du DataFrame "orders"
df_filtre = orders[orders["user_id"] != 0]

#  Sauvegarder le nouveau DataFrame en fichier CSV
df_filtre.to_csv("corrected_orders.csv", index=False)
```

### Vérfication du nouveau csv

```python
corrected_orders = pd.read_csv("corrected_orders.csv")


id_users_in_table_user= users['user_id'].unique().tolist()
id_users_in_table_order= corrected_orders['user_id'].unique().tolist()

# comparaison des deux listes pour trouver les IDs d'utilisateurs présents dans la table "orders" mais pas dans la table "users"
missing_users = set(id_users_in_table_order) - set(id_users_in_table_user)  

if missing_users:
    print("IDs d'utilisateurs présents dans la table 'orders' mais pas dans la table 'users':", missing_users)
else:
    print("Tout est correct : tous les IDs d'utilisateurs dans 'orders' sont présents dans 'users'.")
```

# Visualisation du fichier database.db

```python
import sqlite3
import pandas as pd

# 1. Connexion à la base de données
conn = sqlite3.connect('database.db')

# 2. Récupérer le nom de toutes les tables
tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)

print("Tables dans la base de données :")
print(tables['name'].tolist())

# 3. Boucler sur chaque table pour l'afficher
for table_name in tables['name']:
    print(f"\n=== Table : {table_name} ===")
    
    # Charger et afficher le contenu de la table
    df = pd.read_sql_query(f"SELECT * FROM [{table_name}]", conn)
    display(df.head())  # .head() affiche les 5 premières lignes. Retirez-le pour tout voir.

```

```python
# verification présence de l'id 0 dans la table orders
verif = pd.read_sql_query("SELECT * FROM orders WHERE user_id=0;", conn)
display(verif)
```

```python
# Supprimer les lignes où la valeur est 0 dans la colonne "user_id" dans la table orders de la base de données
conn.execute("DELETE FROM orders WHERE user_id = 0")
```

```python
verif = pd.read_sql_query("SELECT * FROM orders WHERE user_id=0;", conn)
display(verif)
```

### Recherche de quelques utilisateurs pour adapter les questions au llm

```python
best_order = pd.read_sql_query("""
    SELECT
        u.email,
        u.first_name,
        u.last_name,
        o.status,
        COUNT(o.order_id) AS nb_commandes
    FROM `orders` AS o
    JOIN users AS u
        ON o.user_id = u.user_id
    GROUP BY
        u.email,
        u.first_name,
        u.last_name,
        o.status
    ORDER BY nb_commandes DESC;
""", con=conn)

display(best_order)
```
